```python
N = int(input("Ingrese el número de edades al que desea calcular el promedio"))
edades = []
for i in range(N):
  edad = int(input(f"Ingrese la edad {i+1}: "))
  edades.append(edad)
  print(f"El promedio de {i+1} edades es ", sum(edades)/len(edades))
#print("El promedio de edades es ", sum(edades)/len(edades))
```

range(n) --> iniciar en 0, hasta n-1, con paso de 1

range(m,n) --> iniciar en m, hasta n-1, con paso de 1

range(m,n,s) --> iniciar en m, hasta n-1, con paso de s

```python
N = int(input("Ingrese un número mayor a 1: "))
numeros = []
for i in range(N):
  numero = float(input(f"Ingrese el número {i+1}: "))
  if numero > 5:
    numeros.append(numero)

print("El número máximo es: ", max(numeros))
print(f"El número mínimo es: {min(numeros)}")
print("El promedio es: ", sum(numeros)/len(numeros))
```
